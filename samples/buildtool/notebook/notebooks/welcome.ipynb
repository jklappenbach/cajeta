# Welcome to Cajeta notebooks

The first cell builds the session (progress streams below it); every cell after runs warm. Bindings and classes persist across cells, and redefining a class takes effect for later cells.

```cajeta
int32 a = 20;
a + 22;
```

```cajeta
public class Greeter {
    public int32 count;
    public Greeter(int32 count) { this.count = count; }
    public int32 next() { this.count = this.count + 1; return this.count; }
}
Greeter g = heap Greeter(a);
g.next();
```

```cajeta
import cajeta.collection.ArrayList;
ArrayList<int32> xs = heap ArrayList<int32>();
xs.add(g.next());
xs.get(0);
```
